```python
# 복습 - 1 부터 10까지 의 자연수 중에서 홀수랑 , 홀수들의 합
#아래 출력예시
# 1,3,5,7,9
#1~10까지의 홀수의 합 : 25
hap = 0 

for i in range (1,11):
    if i %2 == 1:
        print (i, end=' ,')
        hap += i 
print()
print (f'1~10까지의 홀수의 합 : {hap}')
```

```python
hap = 0 

for i in range (1,11):
    if i %2 == 0:
        continue
    print (i, end=' ,')
    hap += i 
print()
print (f'1~10까지의 홀수의 합 : {hap}')
```
