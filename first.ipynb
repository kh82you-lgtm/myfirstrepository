```python
# 성적처리 프로그램
# 90점 이상이면 'A'
# 80점 이상 90점 미만이면 'B'
# 

score = 70

if score >= 90 :
    result = 'A'
elif score >= 80:
    result = 'B'     
elif score >= 70:
    result = 'C'
elif score >= 60:
    result = 'D'          


print (f"학점은 {result} 입니다")
```

```python

#수학점수, 영어점수
#수학점수와 영어점수가 둘다 90점 이상이면 '참 잘했어요"
#수학점수만 90점 이사이면 '좀 더 노력하세요' 
#둘 다 90점 미만이면 '많이 노력하세요'

ma_score = 90
en_score = 80

if ma_score >= 90 :
    if en_score >= 90:
        print ('참 잘 했어요')
    else:
        print ('좀 더 노력하세요')  
else:
    print ('많이 노력하세요')
```

```python
import random

com = random.randint(1,100)


#반복문
#만약 맞으면 반복문 종료
#아니면 반복하면서 크다, 작다


while True: 
    user = int (input('원하는 숫자 입력:'))
    if user == com: 
        print ('정답입니다!')
        break
    elif user > com:
        print ('너무 커요')
    else:
        print ("너무작아요")
```
